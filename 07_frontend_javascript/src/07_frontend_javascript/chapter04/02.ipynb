### 2. 초기 자바스크립트의 변수 선언: var의 치명적인 결함
---
var a = 100 ;  <- 변수의 선언 / 번역 과정 -> window 객체의 하위 속성으로 추가(웹브라우저 기준) / global 객체의 하위 속성으로 추가 (node.js)


```javascript
var i = 10000; // window.i / global.i = 10000

for (var i = 0; i < 10; i++) { // window.i, global.i = 0 ~ 10
    // 반복 로직
    console.log("for문에 정의된 i:", i);
}

console.log(i); // window.i / global.i = 10
```

#### 2.1 var의 3가지 결함
- 호이스팅 (Hoisting)
- 선언 키워드 생략 허용 (암묵적 전역 변수)
- 함수 레벨 스코프 (Function-level Scope)

```javascript
var i = 10000; //window.ii,global.i = 0 ~ 10
for (var i =0; i < 10 ; i++){//window.ii,global.i = 0 ~ 10
    //반복 로직
    console.log("for문에 정의된 i:",i)
}

console.log(i);
```
