- 디버깅 조절자를 활용하여 변수의 이름과 값을 동시에 확인할 수 있다.

### 필수 1 : 문자열 포매팅 기법과 특징 이해하기
#### 문제 1-1 :  기존 방식과 f-string 출력 비교

```python
# % 연산자 방식
menu_name = 'Americano'
menu_price = 45000

menu = "상품: %s | 가격: %d" 
print(menu %('Americano',45000))
```

```python
# fomat 방식
menu_name = 'Americano'
menu_price = 45000

menu = "상품: {menu_name} | 가격: {menu_price}"
menu.format(menu_name ='Americano',menu_price = 45000)
```

```python
# f-string 방식
menu_name =  'Americano'
menu_price = 45000

print(f"상품:{menu_name} | 가격:{menu_price}")
```

### 필수 2 : 정렬 지시자와 천 단위 쉼표 활용하기
#### 문제 2-1 :  깔끔한 금액 표기 프로그램 작성

```python
price = 15000

print(f"금액:{price:>8,}원")
```

### 심화 1 : 디버깅용 등호 조절자(=) 활용하기
#### 문제 3-1 : 등호(=)를 활용한 데이터 추적

```python
current_status = '정산 진행중'
price =  6500

print(f"디버깅 로그->{current_status=},{price=}")
```
