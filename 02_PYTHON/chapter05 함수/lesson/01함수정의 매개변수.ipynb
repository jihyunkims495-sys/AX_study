### 1. 함수의 기본정의와 작동 방식
---
### 함수의 구조 : 입력 -> 처리 -> 구조
- 함수(Function)는 특정 목적을 가진 코드들을 하나의 독립된 상자처럼 묶어 이름을 붙여둔 '재사용 가능한 코드 구조'입니다.
- 함수는 특정한 입력값을 받아서 정해진 처리를 거친 후 결과값을 내놓는 자판기나 마법 상자와 같습니다. 한 번 잘 만들어둔 함수는 필요할 때마다 이름만 불러서 반복적으로 작동시킬 수 있습니다.

### 함수의 처리 구조
def 함수이름(매개변수,...)  #매개변수는 함수가 일을 할 때 사용할 값을 받아 두는 변수
    처리 부분

    return #함수의 처리 결과를 반환

### 함수를 호출하는 방법
함수 이름(...) -> 엔터

### 함수의 목적
코드의 재활용 목적
- 함수는 이미 준비된 '레시피' 같아요. 

### 함수 명명 규칙
- 함수의 이름은 변수와 마찬가지로 숫자로 시작할 수 없으며, 해당 함수가 어떤 기능을 하는지 명확히 알 수 있도록 동사 위주로 명명하는 것이 좋습니다. 
-  '스네이크 케이스(Snake Case)' 방식으로 권장
- 파이썬 인터프리터는 def 키워드를 식별하는 순간 해당 함수 블록을 메모리에 등록해 두었다가, 호출되는 시점에 비로소 실행 제어권을 넘겨 작동시킵니다.

```python
# 기본 함수 구조도
def calc(x) :    #define : 함수 정의/함수 입력
    y = (x * 2) + 1
    return y     #return : 값 처리

# 함수 호출
calc()
```

### 2. 매개변수(parameter)와 인수(argument)
---
- 매개변수 (Parameter)
    - 함수 호출 시, 입력 부분에서 값을 담아주는 변수
    - 함수를 정의할 때 "외부에서 이런 데이터를 입력받을 거야"라고 미리 설정해두는 '입력 통로의 이름'입니다. 함수의 '선언부' 공간에 위치합니다.
- 인수 (Argument)
    - 호출시 사용된 값
    - 함수를 실제로 작동시킬 때 해당 통로를 통해 전달하는 '실제 데이터 값'입니다. 함수의 '호출부' 공간에서 주입됩니다.

```python
# 정의할 때의 name은 '매개변수'
def welcome(name):
    print(f"{name}님, 환영합니다!")

# 호출할 때의 "민수"는 '인수'
welcome("민수")
welcome("영희")
```

### 3. 위치 인수(positional argument) 와 키워드 인수(keyword arguments)
---

- 위치 인수(Positional Arguments)와 키워드 인수(Keyword Arguments)
    - 함수를 호출할 때 정의된 매개변수의 "순서대로" 인수를 차례대로 주입하는 가장 기본적인 방식
    - 인터프리터가 오직 '위치(순서)'만을 기준으로 데이터를 식별하여 매핑하므로, 순서가 바뀌면 의도치 않은 변수에 값이 대입되는 오작동을 유발할 수 있어 순서를 엄격히 준수해야 합니다.

- 키워드 인수를 활용한 자유로운 제어
    - 인수를 전달할 때 "매개변수명=값" 형태로 이름을 명시하여 전달하는 방식입니다. 이 구조를 사용하면 함수 정의부의 매개변수 배치와 상관없이 원하는 순서로 자유롭게 바꾸어 호출할 수 있습니다. 다만 위치 인수와 함께 사용할 때는 반드시 위치 인수를 먼저 작성해야 합니다. 이 방식을 활용하면 인수의 의미가 명확해져 코드의 가독성이 대폭 향상됩니다.

```python
def introduce_self(name, age):
    print(f"이름: {name}, 나이: {age}")

# 1. 위치 인수를 사용한 호출: 정의된 순서대로 정확히 주입해야 함
introduce_self("홍길동", 30)  # 정상 작동
introduce_self(30, "홍길동")  # 오작동 발생 (이름 공간에 수치 30이 식별됨)

# 2. 키워드 인수를 사용한 호출: 매개변수 이름을 특정하므로 순서가 상관없음
introduce_self(age=25, name="이영희")  # 순서를 뒤바꿔도 안전하게 매핑되어 정상 작동
```

##  개념 요약 키워드
---
- **코드 재사용:** 반복되는 로직을 독립된 함수 구조로 묶어 중복을 제거하고 전체 프로그램의 유지보수 효율을 높이는 핵심 원리입니다.
- **매개변수와 인수:** 함수 내의 입력 통로 이름(매개변수)과 실제 호출 시 전달되는 데이터 값(인수)의 매핑 관계를 식별하는 기초 개념입니다.
- **가변 인수 (*args)*:*** 인수의 개수를 고정하지 않고 유연하게 데이터를 수집하여 튜플 구조로 처리하는 작동 방식입니다.
- **키워드 가변 인수 (**kwargs):** 이름과 값을 쌍으로 지정하여 개수 제한 없이 데이터를 수집하고 딕셔너리 구조로 처리하는 작동 방식입니다.
- **return 문:** 함수의 작동을 즉시 종료하고 최종 결과값을 호출부로 반환하여 전체적인 데이터 흐름을 연결하는 명령어입니다.
- **튜플 반환:** 여러 결과값을 동시에 내보낼 때, 파이썬 인터프리터가 자동으로 튜플 구조로 묶어 반환하고 언패킹으로 각각 식별하게 돕는 효율적인 데이터 전달 기법입니다.

## 용어 사전
---
| **용어** | **영어 표기** | **설명** |
|---|---|---|
| **함수** | Function | 반복적으로 사용되는 특정 코드 덩어리를 독립된 하나의 방으로 묶어두고, 필요할 때마다 이름만 불러서 재사용할 수 있도록 만드는 문법 구조입니다. |
| **매개변수** | Parameter | 함수를 정의(설계)할 때 외부로부터 어떤 데이터 값을 넘겨받을지 미리 선언해 두는 ‘받는 변수(이름표)’입니다. |
| **인수(인자)** | Argument | 설계된 함수를 실제로 호출(실행)할 때 함수의 매개변수 자리에 넘겨주는 ‘실제 데이터 값’입니다. |
| **위치 인수** | Positional Argument | 함수를 호출할 때 매개변수가 정의된 순서에 맞춰 값을 전달하는 가장 기본적인 방식입니다. |
| **기본값 매개변수** | Default Parameter | 함수를 정의할 때 매개변수에 기본값을 미리 지정하는 기능입니다. 인수를 전달하지 않으면 기본값이 사용되며, 기본값 매개변수는 일반 매개변수보다 뒤쪽에 작성해야 합니다. |
| **가변 인수** | Variable Arguments (`*args`) | 함수에 개수 제한 없이 여러 값을 전달받는 문법입니다. 전달된 값들은 함수 내부에서 하나의 튜플(Tuple)로 묶입니다. |
| **키워드 가변 인수** | Keyword Variable Arguments (`**kwargs`) | `키=값` 형태의 인수를 개수 제한 없이 전달받는 문법입니다. 전달된 값들은 함수 내부에서 하나의 딕셔너리(Dictionary)로 묶입니다. |
| **반환문** | Return Statement (`return`) | 함수 내부에서 처리한 결과를 함수 밖으로 돌려보내고 함수 실행을 종료하는 명령어입니다. |
| **복수 데이터 반환** | Multiple Return Values | `return a, b`처럼 여러 결과를 반환하는 방식입니다. 실제로는 여러 값이 하나의 튜플 객체로 묶여 반환됩니다. |
