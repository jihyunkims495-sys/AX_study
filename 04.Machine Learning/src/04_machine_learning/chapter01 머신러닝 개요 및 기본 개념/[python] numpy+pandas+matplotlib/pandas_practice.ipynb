### 3. Pandas 정밀 조회 및 결측치 제어

#### 3.1  loc와 iloc를 활용한 행·열 선택
- loc(location)랑 iloc(index location)는 둘 다 Pandas에서 원하는 행/열을 골라내는 조회 방법
- "큰 데이터 중, 데이터 전처리 때 일부 데이터들을 추출해서 가공하기 전 많이 사용하는 기능입니다."
- loc  → 행/열의 이름(label) 기준으로 선택 / 이름으로 찾기
    - df.loc[행 라벨(이름)]
    - df.loc[행 라벨(이름), 컬럼(열)라벨(이름)]
    - (슬라이싱도 가능!) df.loc[시작행:종료행,[컬럼1,컬럼2]] # 특정 범위의 행과 열을 선택도 가능합니다.
    - df.loc[...,시작컬럼:종료컬럼]
- iloc → 행/열의 위치 번호(integer position) 기준으로 선택 / 몇 번째 위치인지 번호로 찾기
    - loc와 사용방법이 동일하지만, 인덱스 번호
    - 범위를 설정할 경우 종료번호는 포함x (인덱스 규칙 동일하게 적용)
        - (e.g) iloc[0:2,0:2] -> 행/열 모두 2는 포함x

```python
# loc 예시
import pandas as pd

df = pd.DataFrame({
    '나이':[25,30,35,40],
    '점수':[80,95,70,85],
    '등급':['B','A','C','B']
},index=['user1','user2','user3','user4'])

df
```

```python
df.loc['user1']  # 행 이름 선택
type(df.loc['user1']) #타입이 나와요~
```

```python
df.loc['user1','점수'] # 열 이름 선택
```

```python
df.loc['user1':'user3',['나이','등급']] # 행 슬라이싱과 열 추출
```

```python
df.loc['user1':'user2','점수':'등급'] # 행/열 슬라이싱 추출
```

```python
# iloc 예시
df.iloc[0:1,0:3]
```

#### 3.3 결측치(Missing Data) 7대 처리 기법
3.3.1 결측치 탐지 및 현황 요약 (Detection)
- 결측치 = 값이 비어 있거나 누락된 데이터의 위치를 찾는 것

- 데이터프레임의 결측치는 가장 먼저 info() 에서 확인합니다.
- 결측치인지 확인하는 메서드 : isna(),isnull()
    - 결측치다  → True / 결측치 아니다 → False
- [중요!!] True = 1로 인식! / False = 0으로 인식!
    (e.g) True + True = 2 / False + 1 = 1

- 결측값 표현
    - np.nan → NumPy의 결측값
    - pd.NA  → Pandas의 결측값
    - None   → Python의 값 없음

```python
import numpy as np

# 결측치 : np.nan, pd.NA , NONE
raw_df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [pd.NA, 5, 6], 'C': [7, 8, 9]})

raw_df.info()
```

```python
raw_df.isna() # → 비어 있으면 True!!!!
```

```python
raw_df.isna().sum() # 비어있는 곳에 브로드캐스팅!!!!
```

3.3.3 통계적 대푯값 대치 (Imputation)
- 결측치를 그냥 지우지 않고, 그 컬럼을 대표할 만한 값으로 채우는 것
- 통계적 대푯값 3가지
    - 평균 mean()
    - 중앙값 median() 
        - 튀는 값들의 영향을 덜 받으므로 중앙값을 대치하기도 합니다. 평균값을 넣으면 데이터가 희석될 가능성이 크므로~
        - 중앙값은 양끝값보다 가운데 위치를 보는 값
    - 최빈값 mode

- 대표 메서드 
    - fillna() → 결측치를 채운다
    - mean() → 평균값을 구한다

```python
# 예시

df_age = pd.DataFrame({'나이': [20, 25, None, 30, 95]})

median_val = df_age['나이'].median()
print("중앙값:",median_val)

df_age['나이'] = df_age['나이'].fillna(median_val)

df_age
```

3.3.4 시계열 직전/직후 값 대치 (Forward / Backward Fill)
- 결측치를 앞이나 뒤에 있는 값으로 채우는 방법으로, 특히 날짜·시간 순서가 있는 데이터에서 많이 쓴다.
- 반대로 Backward Fill(후방 채우기)은 바로 다음 값을 끌어와서 결측치를 채웁니다.
- 대표 메서드
    - ffill() : 앞쪽에 있는 값으로 대체
    - bfill() : 뒷쪽에 있는 값으로 대체

```python
ts_df = pd.DataFrame({'온도': [18.2, None, None, 21.0]})

ts_df.ffill()
```

```python
ts_df.bfill()
```

3.3.5 선형 보간법 (Linear Interpolation)
- 앞값과 뒷값 사이를 직선처럼 이어서, 중간의 결측값을 추정해서 채우는 방법
- 앞뒤의 실제 값을 이용해서 그 사이 변화가 일정하다고 가정하고 결측값을 계산하는 방법
- 시간 순서가 있고 값이 비교적 연속적으로 변하는 온도, 센서 측정값 같은 데이터 처리에 사용
- 대표 메서드
    - interpolate()

- method='linear'는 “어떤 방식으로 보간할지 지정하는 옵션”
    - linear는 앞뒤 값을 직선처럼 이어서 가운데를 계산한다. 
    - [중요!!!] : interpolate()의 기본값이 원래 linear

```python
speed_df = pd.DataFrame({'속도': [10.0, None, None, 40.0]})


speed_df['속도'] = speed_df['속도'].interpolate(method='linear')
speed_df
```

```python
speed_df.interpolate()
```

- 이상한 데이터를 교체 / 이상한 값을 정리하고 → 계산 가능한 타입으로 바꾸는 과정
- 지저분한 문자열 데이터를 정리한 뒤, 계산 가능한 숫자 데이터로 바꾸는 전처리 과정

#### 3.4 그룹화 집계(groupby)와 .to_numpy() 상호 변환
- Pandas에서 데이터를 요약한 뒤, 필요하면 NumPy 배열로 넘기는 흐름
- 왜 바꾸냐면, Pandas는 표 구조와 라벨을 다루기 편하고, NumPy는 수치 계산에 강하니까 필요에 따라 넘겨 쓰는 거야.

- groupby() :  같은 기준끼리 묶어서 집계 / 같은 범주끼리 묶어서 평균, 합계, 개수 같은 통계를 내는 기능
- .to_numpy() :  Pandas의 Series나 DataFrame을 NumPy 배열로 바꾸는 메서드

- numeric_only=True → 숫자형 컬럼만 평균 계산에 사용
    - .mean(numeric_only=True) : 문자열 컬럼은 제외하고 숫자 컬럼만 평균 내라 / 조건 명시!!!!!!!

##### 구조
---
Pandas로 원본 데이터 정리
↓
groupby()로 그룹별 요약
↓
필요하면 .to_numpy()
↓
NumPy 배열로 수치 계산
---

```python
sales_df = pd.DataFrame({'지점': ['서울', '서울', '부산', '부산'], '매출': [100, 150, 80, 120]})

sales_df
```

```python
group_df = sales_df.groupby('지점').mean(numeric_only=True)
group_df
```

```python
sales_df['매출'].to_numpy(dtype=np.float64)
```

1. NumPy — “숫자를 계산하기 좋은 구조로 만든다”

- NumPy는 숫자 계산용 배열을 다루는 도구
- 왜 굳이 NumPy를 쓰냐? 숫자가 엄청 많아졌을 때 빠르게 계산하기 좋기 때문
- “숫자들이 지금 어떤 모양으로 메모리에 들어 있고, 어느 방향으로 계산할 것인가?”
- bit, byte, int64가 왜 나오냐?
